```python
dec_pts = 1

datasets = ['trec-covid', 'trec-news', 'webis-touche2020']
dataset_names = ['covid', 'news']

metric_l, k_l, metric_str_l = 'recall', 100, 'R@100'
metric_r, k_r, metric_str_r = 'ndcg', 10, 'N@10'
```

```python

# baseline_rows = [
#     "BM25 & 52.1 & 43.2 & 48.7 & 39.4",
# ]
baseline_rows = []


methods = [
     ('GP CE Greedy B=10', 'gp_al/ce/l6/greedy_epsilon/0/batch_af/10'),
     ('GP CE Random B=10', 'gp_al/ce/l6/greedy_epsilon/1/batch_af/10'),
     # ('method_name_2', 'path_to_methods_res_dir_2'),
]

method_rows = []

for i, (method_name, method_path) in enumerate(methods):

    row_cells = [method_name]

    for ds in datasets:
        ds_path = os.path.join('trials/ir', ds, method_path)

        val_l = get_mean_at_k(ds_path, metric_l, k_l)
        val_r = get_mean_at_k(ds_path, metric_r, k_r)

        str_l = '-' if val_l is None else f"{val_l:.{dec_pts}f}"
        str_r = '-' if val_r is None else f"{val_r:.{dec_pts}f}"

        row_cells.extend([str_l, str_r])

    method_rows.append(" & ".join(row_cells))

    # custom horizontal rule after first method
    # if i == 0:
    #     method_rows.append(r"\midrule")


latex_block = make_two_metric_table(
    datasets=datasets,
    dataset_names=dataset_names,
    metric_str_l=metric_str_l,
    metric_str_r=metric_str_r,
    baseline_rows=baseline_rows,
    method_rows=method_rows,
    caption="MAIN",
    label="tab:main",
)

print(latex_block)
```
